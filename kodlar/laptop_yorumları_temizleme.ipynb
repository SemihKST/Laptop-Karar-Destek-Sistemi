```python
import pandas as pd
import re

# =====================================
# DOSYALAR
# =====================================

GIRIS_DOSYASI = "ham_laptop_yorumlari.xlsx"
CIKIS_DOSYASI = "temiz_laptop_yorumlari.xlsx"

df = pd.read_excel(GIRIS_DOSYASI)

print("Ham yorumlar okundu.")
print("Toplam satır:", len(df))


# =====================================
# YORUM TEMİZLEME
# =====================================

def yorum_temizle(yorum):

    if pd.isna(yorum):
        return ""

    yorum = str(yorum)

    # satır sonlarını kaldır
    yorum = yorum.replace("\n", " ")

    # fazla boşlukları temizle
    yorum = re.sub(r"\s+", " ", yorum)

    yorum = yorum.strip()

    return yorum


df["yorum_metni"] = df["yorum_metni"].apply(yorum_temizle)


# =====================================
# BOŞ YORUMLARI SİL
# =====================================

df = df[df["yorum_metni"] != ""]


# =====================================
# ÇOK KISA YORUMLARI SİL
# =====================================

df = df[df["yorum_metni"].str.len() >= 10]


# =====================================
# AYNI ÜRÜNDE AYNI YORUM TEKRARINI SİL
# =====================================

onceki = len(df)

df = df.drop_duplicates(
    subset=["ürün_id", "yorum_metni"]
)

sonraki = len(df)

print("Silinen tekrar sayısı:", onceki - sonraki)


# =====================================
# SADECE GEREKLİ SÜTUNLAR
# =====================================

temiz_df = df[
    [
        "ürün_id",
        "ürün_adı",
        "yorum_metni",
        "yorum_tarihi",
        "ürün_linki"
    ]
]


# =====================================
# KAYDET
# =====================================

temiz_df.to_excel(
    CIKIS_DOSYASI,
    index=False
)

print()
print("Temiz yorum dosyası oluşturuldu.")
print("Dosya:", CIKIS_DOSYASI)
print("Toplam temiz yorum:", len(temiz_df))
```
