```python
import pandas as pd

# ====================================
# DOSYALAR
# ====================================

URUN_DOSYASI = "temiz_laptop_verileri.xlsx"
YORUM_OZELLIK_DOSYASI = "yorum_ozellikleri.xlsx"
DUYGU_DOSYASI = "urun_duygu_ozeti.xlsx"

CIKTI_DOSYASI = "analiz_verisi.xlsx"


# ====================================
# EXCEL DOSYALARINI OKU
# ====================================

urun_df = pd.read_excel(URUN_DOSYASI)
yorum_df = pd.read_excel(YORUM_OZELLIK_DOSYASI)
duygu_df = pd.read_excel(DUYGU_DOSYASI)

print("Ürün dosyası sütunları:")
print(urun_df.columns.tolist())

print("\nYorum özellikleri sütunları:")
print(yorum_df.columns.tolist())

print("\nDuygu analizi sütunları:")
print(duygu_df.columns.tolist())


# ====================================
# ÜRÜN ID KONTROLÜ
# ====================================

if "ürün_id" not in urun_df.columns:
    urun_df.insert(0, "ürün_id", range(1, len(urun_df) + 1))
    print("\nÜrün dosyasına ürün_id eklendi.")


# ====================================
# YORUM ÖZELLİKLERİNİ BİRLEŞTİR
# ====================================

yorum_merge = yorum_df.copy()

if "ürün_adı" in yorum_merge.columns:
    yorum_merge = yorum_merge.drop(columns=["ürün_adı"])

analiz_df = pd.merge(
    urun_df,
    yorum_merge,
    on="ürün_id",
    how="left"
)


# ====================================
# DUYGU ANALİZİNİ BİRLEŞTİR
# ====================================

duygu_merge = duygu_df.copy()

if "ürün_adı" in duygu_merge.columns:
    duygu_merge = duygu_merge.drop(columns=["ürün_adı"])

analiz_df = pd.merge(
    analiz_df,
    duygu_merge,
    on="ürün_id",
    how="left"
)


# ====================================
# BOŞ SAYISAL DEĞERLERİ 0 İLE DOLDUR
# ====================================

sayisal_kolonlar = analiz_df.select_dtypes(
    include=["number"]
).columns

analiz_df[sayisal_kolonlar] = analiz_df[
    sayisal_kolonlar
].fillna(0)


# ====================================
# BOŞ KATEGORİK DEĞERLERİ DOLDUR
# ====================================

kategorik_kolonlar = analiz_df.select_dtypes(
    include=["object"]
).columns

analiz_df[kategorik_kolonlar] = analiz_df[
    kategorik_kolonlar
].fillna("Bulunamadı")


# ====================================
# KAYDET
# ====================================

analiz_df.to_excel(
    CIKTI_DOSYASI,
    index=False
)

print("\nBirleştirme tamamlandı.")
print("Toplam ürün sayısı:", len(analiz_df))
print("Dosya oluşturuldu:", CIKTI_DOSYASI)
```
